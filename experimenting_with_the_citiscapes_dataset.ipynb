```python
import torch

from torchvision.io import read_image

import matplotlib.pyplot as plt
```

```python
msk = read_image("/content/aachen_000000_000019_gtFine_labelIds.png")
```

```python
msk.shape
```

```python
torch.unique(msk)
```

```python
plt.imshow(msk.permute(1, 2, 0), cmap="gray")
```

```python
original_id_to_cityscapes_id = {
    7: 0,
    8: 1,
    11: 2,
    12: 3,
    13: 4,
    17: 5,
    19: 6,
    20: 7,
    21: 8,
    22: 9,
    23: 10,
    24: 11,
    25: 12,
    26: 13,
    27: 14,
    28: 15,
    31: 16,
    32: 17,
    33: 18
}
```

```python
for original_id, cityscapes_id in original_id_to_cityscapes_id.items():
  print(f"{original_id} : {cityscapes_id}")
```

```python
msk_processed = torch.full((1, msk.shape[1], msk.shape[2]), 255)
```

```python
for original_id, cityscapes_id in original_id_to_cityscapes_id.items():
  msk_processed[msk == original_id] = cityscapes_id
```

```python
plt.imshow(msk_processed.permute(1, 2, 0), cmap="gray")
```
