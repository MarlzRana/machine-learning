```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df_1 = pd.read_csv('/content/drive/My Drive/3yp/logs/non-synth/pretrain_modresnet50_unet.csv')
df_2 = pd.read_csv('/content/drive/My Drive/3yp/logs/non-synth/pretrain_modresnet34_unet.csv')

# Plot
plt.figure(figsize=(10, 5))
plt.plot(df_1['Epoch Number'], df_1['Train Loss'], label='ResNet50 Training Loss')
plt.plot(df_1['Epoch Number'], df_1['Val Loss'], label='ResNet50 Validation Loss')

plt.plot(df_2['Epoch Number'], df_2['Train Loss'], label='ResNet34 Training Loss')
plt.plot(df_2['Epoch Number'], df_2['Val Loss'], label='ResNet34 Validation Loss')

# Adding title and labels
plt.title('Training and Validation Loss per Epoch for ResNet-34-U-Net-B & ResNet-50-U-Net-B')
plt.xlabel('Epoch Number')
plt.ylabel('Weighted Dice Loss')

# Add a legend to the graph
plt.legend()

plt.xlim(10, max(df_1['Epoch Number'].max(), df_2['Epoch Number'].max()))

plt.ylim(0.2, 0.325)


# Show the plot
plt.show()
```
